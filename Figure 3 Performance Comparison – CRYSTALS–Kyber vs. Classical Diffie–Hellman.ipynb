```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Simulated execution times in milliseconds (these values are approximate)
num_keys = np.array([1, 10, 100, 500, 1000, 5000, 10000])  # Number of key exchanges
dh_times = np.array([0.5, 5, 50, 250, 500, 2500, 5000])    # Classical DH runtimes
kyber_times = dh_times * 2.3  # Kyber takes ~2.3× longer

# Plot the data
plt.figure(figsize=(8, 5))
sns.lineplot(x=num_keys, y=dh_times, marker='o', label='Classical Diffie–Hellman', color='blue')
sns.lineplot(x=num_keys, y=kyber_times, marker='s', label='CRYSTALS–Kyber', color='red')

# Labels and Titles
plt.xlabel("Number of Key Exchanges")
plt.ylabel("Runtime (ms)")
plt.title("Performance Comparison – CRYSTALS–Kyber vs. Classical Diffie–Hellman")
plt.xscale('log')  # Use a logarithmic scale for better visualization
plt.yscale('log')
plt.grid(True)
plt.legend()

# Save the figure in high resolution
plt.savefig('kyber_vs_dh.png', dpi=300)
plt.show()
```
