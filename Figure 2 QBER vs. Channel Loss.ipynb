```python
import numpy as np
import matplotlib.pyplot as plt

# Function to simulate QBER under different channel losses
def simulate_qkd_qber(channel_losses):
    qber_values = []
    for loss in channel_losses:
        # Simulate errors (random noise + loss effects)
        error_rate = np.random.uniform(0.01, 0.05) + (loss / 100)  # Assume QBER increases with loss
        qber_values.append(min(error_rate, 1))  # Ensure QBER does not exceed 100%
    
    return qber_values

# Define channel loss values (in dB)
channel_losses = np.linspace(0, 20, 20)  # Simulating loss from 0dB to 20dB
qber_values = simulate_qkd_qber(channel_losses)

# Plot QBER vs. Channel Loss
plt.figure(figsize=(8, 5))
plt.plot(channel_losses, qber_values, marker='o', linestyle='-', color='b', label='QBER')
plt.axhline(y=0.05, color='r', linestyle='--', label='5% QBER threshold')
plt.xlabel('Channel Loss (dB)')
plt.ylabel('Quantum Bit Error Rate (QBER)')
plt.title('QBER vs. Channel Loss in QKD')
plt.legend()
plt.grid(True)

# Save the figure
plt.savefig('qber_vs_loss.png', dpi=300)  # Save in high resolution for journal
plt.show()
```
